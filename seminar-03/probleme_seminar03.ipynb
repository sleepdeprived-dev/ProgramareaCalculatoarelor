```python
# Structura alternativă – if / elif / else:
# TechRo SRL primește o livrare. Cantitatea recepționată nu poate fi negativă,
# iar prețul unitar trebuie să fie strict pozitiv. Identifică expresia logică pentru
# date valide.
# Scrie un program care citește cele două valori și afișează «Date valide» sau
# «Date invalide»

try:
    pret_unitar = float(input("Introdu pretul unitar: "))
    cantitate = int(input("Introdu cantitatea: "))

    if pret_unitar > 0 and cantitate >= 0:
        print("Date valide")
    else:
        print("Date invalide")

except ValueError:
    print("Trebuie sa introduci numere.")
```

```python
# Structura repetitivă – for (1):
# Un operator dorește să simuleze recepția a 5 colete. Pentru fiecare colet, afișează
# numărul coletului.
# Extensie: citește cantitatea din fiecare colet și calculează cantitatea totală primită. Folosește o variabilă acumulator.

total = 0

for package_number in range(1, 6):
    print(f"Colet: {package_number}")

    try:
        cantitate = int(input("Cantitatea: "))
        total += cantitate
        print(f"Cantitatea totala primita este: {total}")

    except ValueError:
        print("Trebuie sa fie un numar.")
```

```python
# Structura repetitivă – for (2):
# O firmă dorește să introducă cheltuielile pentru fiecare dintre cele 12 luni
# și să calculeze cheltuiala totală anuală și media lunară.

total = 0

for luna in range(1, 13):
    try:
        cheltuiala = float(input("Introdu cheltuiala: "))
        total += cheltuiala
        media = total/12

        print(f"Cheltuiala totala anuala: {total}")
        print(f"Media lunara: {media}")

    except ValueError:
        print("Trebuie sa introduci un numar.")
```

```python
# Structura repetitivă – while (1):
# Un magazin are 23 de produse de ambalat. În fiecare pas sunt ambalate cel mult 5
# produse. Simulează procesul până când nu mai rămân produse. Afișează numărul
# pasului și produsele rămase.

produse = 23
pas = 0

while produse > 0:
    pas += 1
    produse -= 5

    if produse < 0:
        produse = 0

    print(f"Pasul {pas} | Produse ramase: {produse}")
```

```python
# Structura repetitivă – while (2):
# O firmă are ca obiectiv vânzări de cel puțin 10 000 lei. În fiecare zi se introduce valoarea
# vânzărilor, iar programul continuă până când suma cumulată atinge sau depășește
# obiectivul. La final se afișează numărul de zile necesare

suma = 0
zile = 0

while suma < 10000:
    try:
        vanzari = float(input("Introduceti vanzarile din ziua curenta: "))
        suma += vanzari
        zile += 1

        print(f"Numarul de zile necesare este: {zile}")

    except ValueError:
        print("Trebuie sa introduci un numar.")
        break
```

```python
# Structura repetitivă – while (3):
# O firmă are o datorie de 20 000 lei. În fiecare lună achită o anumită sumă, introdusă de
# utilizator. Programul continuă până când datoria este achitată integral și afișează câte
# luni au fost necesare.

datorie = 20000
luni = 0

while datorie > 0:
    try:
        suma = float(input("Introduceti suma achitata: "))
        datorie -= suma
        luni += 1

        print(f"Numarul de luni necesare este: {luni}")

    except ValueError:
        print("Trebuie sa introduci un numar.")
        break
```

```python
# Exercitiul 1
# Citiți un număr întreg și afișați dacă este pozitiv, negativ sau zero

try:
    num = int(input("Introdu un numar: "))
    if num == 0:
        print("Zero")
    elif num < 0:
        print("Negativ")
    elif num > 0:
        print("Pozitiv")

except ValueError:
    print("Trebuie sa fie un numar intreg.")
```

```python
# Exercitiul 2
# Citiți două numere și afișați valoarea maximă. Tratați explicit cazul în care sunt egale.

try:
    x = float(input("Introdu primul numar: "))
    y = float(input("Introdu al doilea numar: "))

    if x == y:
        print("Ambele numere sunt egale")
    else:
        max(x, y)

except ValueError:
    print("Trebuie sa fie numere.")
```

```python
# Exercitiul 3
# Citiți trei numere și afișați-le în ordine crescătoare folosind instrucțiuni condiționale.

try:
    x = float(input("Primul numar: "))
    y = float(input("Al doilea numar: "))
    z = float(input("Al treilea numar: "))

    numere = [x, y, z]
    numere.sort()
    print(numere)

except ValueError:
    print("Trebuie sa fie numere.")
```

```python
# Exercitiul 4
# Se citește nota unui student. Afișați: „Respins”, „Promovat”, „Bine”, „Foarte bine” sau „Excelent”, în funcție de intervalul notei.

try:
    nota = float(input("Introdu nota: "))

    if nota < 1 or nota > 10:
        print("Notele pot fi doar intre 1 si 10")
    elif 1 <= nota < 5:
        print("Respins")
    elif 5 <= nota < 7:
        print("Promovat")
    elif 7 <= nota < 8:
        print("Bine")
    elif 8 <= nota < 9.50:
        print("Foarte bine")
    elif 9.50 <= nota <= 10:
        print("Excelent")

except ValueError:
    print("Trebuie sa fie un numar.")
```

```python
# Exercitiul 5
# Realizați un meniu simplu cu match: 1 – adunare, 2 – scădere, 3 – înmulțire, 4 – împărțire.

try:
    x = float(input("Primul numar: "))
    y = float(input("Al doilea numar: "))

    print("""
    Calculator:
    1. Adunare
    2. Scadere
    3. Inmultire
    4. Impartire
    """)
    option = int(input("Introdu o optiune: "))

    match option:
        case 1:
            print(f"{x} + {y} = {x+y}")
        case 2:
            print(f"{x} - {y} = {x-y}")
        case 3:
            print(f"{x} * {y} = {x*y}")
        case 4:
            if y == 0:
                print(f"{x} / {y} = Nedefinit")
            else:
                print(f"{x} / {y} = {x/y}")

except ValueError:
    print("Trebuie sa introduci numere.")
```

```python
# Exercitiul 6
# Afișați numerele de la 1 la n folosind for, apoi folosind while.

try:
    n = int(input("Introdu un numar: "))

    for i in range(1, n + 1):
        print(i, end=" ")

    print()

    pas = 1

    while pas <= n:
        print(pas, end=" ")
        pas += 1

except ValueError:
    print("Trebuie sa introduci un numar.")
```

```python
# Structura alternativă – if / elif / else:
# TechRo SRL primește o livrare. Cantitatea recepționată nu poate fi negativă,
# iar prețul unitar trebuie să fie strict pozitiv. Identifică expresia logică pentru
# date valide.
# Scrie un program care citește cele două valori și afișează «Date valide» sau
# «Date invalide»

try:
    pret_unitar = float(input("Introdu pretul unitar: "))
    cantitate = int(input("Introdu cantitatea: "))

    if pret_unitar > 0 and cantitate >= 0:
        print("Date valide")
    else:
        print("Date invalide")

except ValueError:
    print("Trebuie sa introduci numere.")
```

```python
# Structura repetitivă – for (1):
# Un operator dorește să simuleze recepția a 5 colete. Pentru fiecare colet, afișează
# numărul coletului.
# Extensie: citește cantitatea din fiecare colet și calculează cantitatea totală primită. Folosește o variabilă acumulator.

total = 0

for package_number in range(1, 6):
    print(f"Colet: {package_number}")

    try:
        cantitate = int(input("Cantitatea: "))
        total += cantitate
        print(f"Cantitatea totala primita este: {total}")

    except ValueError:
        print("Trebuie sa fie un numar.")
```

```python
# Structura repetitivă – while (1):
# Un magazin are 23 de produse de ambalat. În fiecare pas sunt ambalate cel mult 5
# produse. Simulează procesul până când nu mai rămân produse. Afișează numărul
# pasului și produsele rămase.

produse = 23
pas = 0

while produse > 0:
    pas += 1
    produse -= 5

    if produse < 0:
        produse = 0

    print(f"Pasul {pas} | Produse ramase: {produse}")
```

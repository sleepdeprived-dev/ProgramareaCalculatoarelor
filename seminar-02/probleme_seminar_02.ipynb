```python
# Problema 1: Cititi doua numere a și b și afișați suma, diferența, produsul, media, câtul, restul și pătratul numerelor.

try:
    a = float(input("Scrie a: "))
    b = float(input("Scrie b: "))
    medie = (a+b)/2
    print(f"{a} + {b} = {a+b}")
    print(f"{a} - {b} = {a-b}")
    print(f"{a} * {b} = {a*b}")

    if b != 0:
        print(f"{a} / {b} = {a/b}")
        print(f"{a} % {b} = {a%b}")
    elif b == 0:
        print(f"{a} / {b} = Nedefinit, impartirea la 0 este imposibila.")
        print(f"{a} % {b} = Nedefinit.")

    print(f"{a}^2 = {a**2}")
    print(f"{b}^2 = {b**2}")
    print(f"Media dintre {a} si {b} este {medie}")

except ValueError:
    print("Trebuie sa introduci un numar.")
```

```python
# Problema 2: Se citește prețul unui produs fără TVA. Să se calculeze prețul final, știind că TVA-ul este 19%.

try:
    pret = float(input("Scrie pretul initial: "))

    if pret < 0:
        print("Pretul nu poate fi un numar negativ.")
    elif pret >= 0:
        pret_cu_tva = pret + (19/100 * pret)
        print(f"Pretul initial: {pret} RON")
        print(f"Pretul cu TVA inclus: {pret_cu_tva} RON")

except ValueError:
    print("Trebuie sa fie un numar.")
```

```python
# Problema 3: Se citește un număr de minute m. Să se transforme în ore și minute.

try:
    minute = int(input("Scrie numarul de minute: "))

    if minute < 0:
        print("Minutele nu pot fi negative.")
    elif minute >= 0:
        ore = minute//60
        minute_ramase = minute % 60
        print(f"{ore} ore si {minute_ramase} minute")

except ValueError:
    print("Trebuie sa fie un numar.")
```

```python
# Problema 4: Se citește un număr natural de două cifre. Să se calculeze suma cifrelor sale.

try:
    x = int(input("Introdu un numar natural de doua cifre: "))

    if 10 <= x <= 99:
        suma = x // 10 + x % 10
        print(f"Suma cifrelor este {suma}.")

    else:
        print("Trebuie sa fie un numar natural de doua cifre.")

except ValueError:
    print("Trebuie sa fie un numar.")

```

```python
# Problema 5: Se citesc:
# - vârsta clientului;
# - dacă are card de fidelitate.
# Clientul primește reducere dacă are sub 18 ani sau peste 60 de ani sau dacă are card de fidelitate.
# Citiți datele de la tastatură, construiți expresia logică și afișați rezultatul True sau False.

try:
    varsta = int(input("Scrie varsta clientului: "))

    if varsta < 0:
        print("Varsta nu poate fi negativa.")

    else:
        areCard = input("Are card de fidelitate? (da/nu): ")

        if areCard.lower().strip() == "da":
            areCard = True

        elif areCard.lower().strip() == "nu":
            areCard = False

        else:
            print("Raspunde cu da sau nu.")

        areReducere = varsta < 18 or varsta > 60 or areCard
        print(f"Are reducere: {areReducere}")

except ValueError:
    print("Trebuie sa introduci un numar.")
```
